#### Imports

```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import mlflow 
import os
```

```python
df = pd.read_csv("../dataset/dataset_treino_knn.csv")
df.head()
```

```python
mlflow.set_tracking_uri("http://127.0.0.1:5000")
mlflow.set_experiment("KNN Pose Classifier")

with mlflow.start_run(run_name="Experiment 1"):
    X = df.drop(columns=['label', 'exercicio'])
    y = df['label']

    X_train, y_train, X_test, y_test = train_test_split(X, y, random_state=42, stratify=y, test_size=0.2)

    model = KNeighborsClassifier(n_neighbors = 5)
    model.fit(X, y)

    mlflow.sklearn.log_model(model, "knn_model")
    print("Treino concluído!")
```
